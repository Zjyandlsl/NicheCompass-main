```python
import anndata as ad
import pandas as pd

# ==========================================
# 1. 配置文件路径并加载数据
# ==========================================
# 注意：这里使用了你上一条指令中实际覆盖保存的路径
file_path = "/home/zhangjunyi/xiangmu/nichecompass-main/datasets/Human_breast_cancer/Human_breast_cancer_ViHBC/Human_breast_cancer_ViHBC_NicheCompassAndmy_Final.h5ad"

print(f"正在加载精简后的 h5ad 文件...\n路径: {file_path}")
adata = ad.read_h5ad(file_path)
obs_df = adata.obs

# ==========================================
# 2. 智能提取各类别列名
# ==========================================
# 类别1: 原始病理注释 (通常是 annot_type)
base_cols = ['annot_type'] if 'annot_type' in obs_df.columns else []

# 类别3 & 4: 我们的核心成果 (宏观与微观生态位)
level_cols = ['Level1_Macro_Region', 'Level2_Micro_Niche']

# 类别2: SpaCET 反卷积分数 (动态识别：排除已知列和坐标列后的浮点数分类)
exclude_cols = base_cols + level_cols + ['in_tissue', 'array_row', 'array_col', 'x', 'y', 'batch']
spacet_cols = [c for c in obs_df.columns if c not in exclude_cols and obs_df[c].dtype in ['float64', 'float32']]

# 为了打印排版美观，我们只挑选最具代表性的 4 个 SpaCET 细胞分数展示
spacet_display = [c for c in ['Malignant', 'CAF', 'Macrophage', 'Endothelial'] if c in spacet_cols]
# 如果没找到这些特定的，就随便取前4个
if not spacet_display:
    spacet_display = spacet_cols[:4]

# 合并所有要展示的列
display_cols = base_cols + spacet_display + level_cols

# ==========================================
# 3. 格式化打印展示 (前 5 个 Spot)
# ==========================================
print("\n" + "="*80)
print("🎉 最终精简版 h5ad 文件前 5 个 Spot 的核心信息概览 🎉")
print("="*80)

# 提取前 5 行数据
head_5 = obs_df[display_cols].head(5)

# 逐行打印，带有清晰的中文注释
for spot_id, row in head_5.iterrows():
    print(f"\n🟢 【Spot ID】: {spot_id}")
    
    print("  ├─ 1. 原始基础列及病理注释")
    annot = row['annot_type'] if 'annot_type' in row else "未知"
    print(f"  │    └─ annot_type (病理学家标注): {annot}")
    
    print("  ├─ 2. SpaCET 反卷积细胞成分分数 (部分展示)")
    for sp_col in spacet_display:
        print(f"  │    ├─ {sp_col}: {row[sp_col]:.4f}")
    print("  │    └─ ... (包含其余所有类型细胞)")
    
    print("  ├─ 3. 最终简略版生态位 (Level 1)")
    print(f"  │    └─ Level1_Macro_Region: {row['Level1_Macro_Region']}  <-- [肿瘤/健康二分类]")
    
    print("  └─ 4. 最终详细版生态位 (Level 2)")
    print(f"       └─ Level2_Micro_Niche: {row['Level2_Micro_Niche']}  <-- [单轴/多轴/休眠状态]")
    
print("\n" + "="*80)
print(f"💡 提示：adata.obs 现在的总列数为 {obs_df.shape[1]} 列，极其轻量纯净！")
print("="*80)

# 如果是在 Jupyter 中，额外打印一个 DataFrame 表格供直观浏览
display(head_5)
```
