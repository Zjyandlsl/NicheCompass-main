```python
# -*- coding: utf-8 -*-
import json
# ===================== 批量填入你的多个文件路径 =====================
# 直接在列表中添加所有需要转换的 ipynb 文件路径
NOTEBOOK_FILES = [
    "/home/zhangjunyi/xiangmu/nichecompass-main/TuMeNiche/src/my/case/parameter_analysis/n_sampled_neighbors/myAndSpaCET_Human_breast_cancer_ViHBC.ipynb",
    "/home/zhangjunyi/xiangmu/nichecompass-main/TuMeNiche/src/my/case/parameter_analysis/n_sampled_neighbors/myAndSpaCET_Human_Prostate_Cancer.ipynb",
    "/home/zhangjunyi/xiangmu/nichecompass-main/TuMeNiche/src/my/case/parameter_analysis/n_sampled_neighbors/myAndSpaCET_SN048_A121573_Rep1.ipynb"
    # 往下继续添加，每行一个，逗号分隔
    # "/路径/文件2.ipynb",
    # "/路径/文件3.ipynb",
]
# ====================================================================

def convert_single_notebook(ipynb_path):
    """处理单个ipynb文件，转成py"""
    try:
        # 自动生成输出路径（替换后缀）
        py_path = ipynb_path.replace(".ipynb", ".py")
        
        # 读取文件
        with open(ipynb_path, 'r', encoding='utf-8') as f:
            notebook = json.load(f)

        all_code = []
        # 提取代码单元格
        for cell in notebook["cells"]:
            if cell["cell_type"] == "code":
                code = "".join(cell["source"])
                all_code.append(code)

        # 写入py文件
        with open(py_path, 'w', encoding='utf-8') as f:
            f.write("# 自动提取自 Jupyter Notebook\n")
            f.write(f"# 源文件：{ipynb_path}\n\n")
            f.write("\n\n".join(all_code))

        print(f"✅ 处理完成：{ipynb_path}")
        print(f"💾 生成文件：{py_path}\n")

    except FileNotFoundError:
        print(f"❌ 错误：未找到文件 → {ipynb_path}\n")
    except Exception as e:
        print(f"❌ 处理失败 {ipynb_path}：{str(e)}\n")

def batch_convert_notebooks():
    """批量转换所有文件"""
    print(f"📌 开始批量转换，共 {len(NOTEBOOK_FILES)} 个文件\n")
    for file_path in NOTEBOOK_FILES:
        convert_single_notebook(file_path)
    print("🎉 所有文件处理完毕！")

if __name__ == "__main__":
    batch_convert_notebooks()
```

```python
# 导入依赖库
import os
import scanpy as sc

def check_he_image_in_h5ad(file_path: str):
    """
    检测h5ad文件中是否包含H&E图像
    :param file_path: h5ad文件的绝对路径
    """
    # 1. 检查文件是否存在
    if not os.path.exists(file_path):
        print(f"❌ 错误：文件不存在！路径：{file_path}")
        return

    # 2. 读取h5ad文件（AnnData对象）
    print(f"✅ 正在读取文件：{file_path}")
    adata = sc.read_h5ad(file_path)
    print(f"✅ 文件读取成功！数据维度：{adata.shape}")

    # 3. 定义H&E图像的关键词（匹配空间转录组标准命名）
    he_keywords = ["he", "H&E", "hematoxylin", "histology", "image", "images", "spatial"]
    found_he = False
    detect_details = []

    # --------------------------
    # 核心检测逻辑（标准存储位置）
    # --------------------------
    # 位置1：uns（未结构化数据）→ 空间转录组H&E图像**最标准**的存储位置
    print("\n🔍 开始检测H&E图像...")
    # 优先检测标准空间图像路径：adata.uns['spatial'][样本ID]['images']['hires']
    if "spatial" in adata.uns:
        spatial_data = adata.uns["spatial"]
        for sample_id, sample_info in spatial_data.items():
            if isinstance(sample_info, dict) and "images" in sample_info:
                # 检测高分辨率/低分辨率H&E图像
                if "hires" in sample_info["images"] or "lowres" in sample_info["images"]:
                    found_he = True
                    detect_details.append(f"✅ 找到H&E图像 | 样本ID: {sample_id} | 存储位置: uns['spatial']['{sample_id}']['images']")

    # 位置2：通用检测（兼容非标准存储的H&E数据）
    # 遍历uns中所有键，匹配H&E相关关键词
    for key in adata.uns.keys():
        key_lower = str(key).lower()
        if any(kw in key_lower for kw in [k.lower() for k in he_keywords]):
            # 排除非图像数据
            if not isinstance(adata.uns[key], (str, int, float)):
                found_he = True
                detect_details.append(f"✅ 找到H&E相关数据 | 存储位置: uns['{key}']")

    # 位置3：obsm（观测矩阵）→ 兼容特殊存储格式
    for key in adata.obsm.keys():
        key_lower = str(key).lower()
        if any(kw in key_lower for kw in [k.lower() for k in he_keywords]):
            found_he = True
            detect_details.append(f"✅ 找到H&E相关数据 | 存储位置: obsm['{key}']")

    # --------------------------
    # 输出最终结果
    # --------------------------
    print("\n" + "-"*50)
    if found_he:
        print("🎉 检测结果：文件中 **包含H&E图像**！")
        for detail in detect_details:
            print(detail)
    else:
        print("❌ 检测结果：文件中 **未找到H&E图像**")
    print("-"*50)

# --------------------------
# 主程序：替换为你的文件路径
# --------------------------
if __name__ == "__main__":
    # 你的h5ad文件绝对路径
    h5ad_file = "/home/zhangjunyi/xiangmu/nichecompass-main/datasets/12_colorectal_cancer_CRC/12_colorectal_cancer_CRC_h5ad/SN048_A121573_Rep1.h5ad"
    # 执行检测
    check_he_image_in_h5ad(h5ad_file)
```

```python
import anndata as ad
import pandas as pd
import os

# ===================== 核心配置 =====================
# h5ad 文件所在文件夹路径
H5AD_DIR = "/home/zhangjunyi/xiangmu/nichecompass-main/datasets/12_colorectal_cancer_CRC/12_colorectal_cancer_CRC_h5ad"
# 病理注释列名
ANNOT_COL = "pathology_annotation"
# 定义无效/需要剔除的标签（和你之前的逻辑完全一致）
INVALID_TAGS = ['exclude', 'nan', 'NaN', 'None', '', 'NA']
# =====================================================

# 1. 获取文件夹内所有有效的 h5ad 文件
h5ad_files = [
    f for f in os.listdir(H5AD_DIR) 
    if f.endswith('.h5ad') and not f.startswith('._')  # 排除mac系统临时文件
]
h5ad_paths = [os.path.join(H5AD_DIR, f) for f in h5ad_files]

print(f"=== 批量清洗任务启动 ===")
print(f"目标文件夹: {H5AD_DIR}")
print(f"待处理文件总数: {len(h5ad_paths)} 个\n")

# 2. 循环处理每个文件
for idx, path in enumerate(h5ad_paths, 1):
    file_name = os.path.basename(path)
    print(f"{'='*60}")
    print(f"【文件 {idx}/{len(h5ad_paths)}】正在处理: {file_name}")
    print(f"文件路径: {path}")
    print(f"{'='*60}")

    try:
        # 读取原始 h5ad 文件
        adata = ad.read_h5ad(path)
        print(f"✅ 文件读取成功 | 原始 Spot 数量: {adata.n_obs}")

        # 检查是否存在病理注释列
        if ANNOT_COL not in adata.obs.columns:
            print(f"⚠️  警告：文件中未找到 '{ANNOT_COL}' 列，跳过清洗，直接保存原文件")
            adata.write_h5ad(path)
            continue

        # 清洗逻辑（和你之前的代码完全一致）
        raw_annots = adata.obs[ANNOT_COL].astype(str).str.strip()
        valid_mask = (adata.obs[ANNOT_COL].notna()) & (~raw_annots.isin(INVALID_TAGS))

        initial_count = adata.n_obs
        adata_clean = adata[valid_mask].copy()
        removed_count = initial_count - adata_clean.n_obs

        # 打印清洗结果
        print(f"🧹 清洗完成：")
        print(f"   - 原始 Spot 数量: {initial_count}")
        print(f"   - 剔除无效/未标注 Spot 数量: {removed_count}")
        print(f"   - 最终保留有效 Spot 数量: {adata_clean.n_obs}")

        if removed_count == 0:
            print("✅ 该切片所有 Spot 均有有效病理标注，无需剔除。")

        # 覆盖保存清洗后的文件
        adata_clean.write_h5ad(path)
        print(f"✅ 已覆盖保存清洗后的文件：{file_name}")

    except Exception as e:
        print(f"❌ 处理文件 {file_name} 时出错: {str(e)}")
        continue

print(f"\n{'='*60}")
print("🎉 所有文件处理完成！")
print(f"共成功处理 {len(h5ad_paths)} 个 h5ad 文件")
```
